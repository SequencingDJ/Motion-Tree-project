Pr (pre 2015), Motion Tree's were constructed using two individual proteins, subjecting them to clusterings and then calculating the distance difference matrix (Dpair).
Dab (pair) = magnitude {d(ab,1)-(dab,2)}
where d denotes the difference between atoms a and b of proteins 1 and 2.

For MD siumulations, the Motion Tree was extended to look specifically at one protein only. Thus, Dpair was replaced with the standard deviation in distance fluctuations in the structual ensemble. 

Dab = <delta d^2 mn>^1/2
